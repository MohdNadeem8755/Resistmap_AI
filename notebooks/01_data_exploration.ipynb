```python
import pandas as pd

file_path = r"D:\ResistMap-AI\data\raw\isolates.tsv"
df = pd.read_csv(file_path, sep="\t", low_memory=False)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("\nColumn names:")
print(df.columns.tolist())

display(df.head())
```

```python
# Validate the current dataset without changing it
required = ["Isolate", "BioSample", "AMR genotypes", "Location"]
missing_columns = [col for col in required if col not in df.columns]

print("Missing core columns:", missing_columns)
print("Exact duplicate rows:", df.duplicated().sum())

for col in required:
    if col in df.columns:
        values = df[col].astype("string").str.strip()
        missing = values.isna() | values.fillna("").eq("")
        print(f"{col}: {missing.sum():,} blank/missing values")

print("\nMost common organism groups:")
display(df["#Organism group"].value_counts(dropna=False).head(15))

print("\nRepeated non-empty isolate IDs:")
ids = df["Isolate"].astype("string").str.strip()
valid_ids = ids[ids.notna() & ids.ne("")]
print(valid_ids.duplicated().sum())

print("\nFive complete AMR result strings:")
for value in df["AMR genotypes"].dropna().head(5):
    print(value)
    print()
```
